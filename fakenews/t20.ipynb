```python
import pandas as pd
import matplotlib.pyplot as plt
import textstat

# 1. Load dataset
df = pd.read_csv("fakenews.csv")

# 2. Handle missing values
df["headline"] = df["headline"].fillna("").astype(str)
df["body_text"] = df["body_text"].fillna("").astype(str)

# Combine headline and body text
df["text"] = df["headline"] + " " + df["body_text"]

# 3. Calculate readability scores
results = []

for index, row in df.iterrows():
    text = row["text"].strip()

    if text:
        flesch_reading_ease = textstat.flesch_reading_ease(text)
        flesch_kincaid_grade = textstat.flesch_kincaid_grade(text)
        gunning_fog = textstat.gunning_fog(text)
    else:
        flesch_reading_ease = 0
        flesch_kincaid_grade = 0
        gunning_fog = 0

    results.append({
        "news_id": row.get("news_id", index),
        "label": row["label"],
        "flesch_reading_ease": flesch_reading_ease,
        "flesch_kincaid_grade": flesch_kincaid_grade,
        "gunning_fog_index": gunning_fog
    })

# 4. Create results DataFrame
readability_df = pd.DataFrame(results)

# Display individual article scores
print("Readability Scores:")
print(readability_df.head(10))

# 5. Compare Real and Fake news
comparison = readability_df.groupby("label")[
    [
        "flesch_reading_ease",
        "flesch_kincaid_grade",
        "gunning_fog_index"
    ]
].mean()

print("\nReadability Comparison: Real vs Fake News")
print(comparison.round(2))

# 6. Visualize readability scores
for column in [
    "flesch_reading_ease",
    "flesch_kincaid_grade",
    "gunning_fog_index"
]:
    comparison[column].plot(
        kind="bar",
        figsize=(7, 4)
    )

    plt.title(f"{column.replace('_', ' ').title()}: Real vs Fake News")
    plt.xlabel("News Category")
    plt.ylabel("Average Score")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()

# 7. Save results
readability_df.to_csv("T20_readability_scores.csv", index=False)
comparison.to_csv("T20_readability_comparison.csv")

print("\nTask 20 completed successfully!")
print("Files saved:")
print("1. T20_readability_scores.csv")
print("2. T20_readability_comparison.csv")
```
