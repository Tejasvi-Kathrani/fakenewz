```python
import pandas as pd
import matplotlib.pyplot as plt
import re
from collections import Counter
import numpy as np

# 1. Load dataset
df = pd.read_csv("fakenews.csv")

# 2. Handle missing values
df["headline"] = df["headline"].fillna("").astype(str)
df["body_text"] = df["body_text"].fillna("").astype(str)

# Combine headline and body text
df["text"] = df["headline"] + " " + df["body_text"]

# 3. Calculate lexical diversity for each article
results = []

for index, row in df.iterrows():

    # Convert text to lowercase and extract words
    words = re.findall(r"\b[a-zA-Z]+\b", row["text"].lower())

    total_words = len(words)
    unique_words = len(set(words))

    # Count word frequencies
    word_counts = Counter(words)

    # Type-Token Ratio (TTR)
    ttr = unique_words / total_words if total_words else 0

    # Root Type-Token Ratio (RTTR)
    rttr = unique_words / np.sqrt(total_words) if total_words else 0

    # Corrected Type-Token Ratio (CTTR)
    cttr = (
        unique_words / np.sqrt(2 * total_words)
        if total_words else 0
    )

    # Hapax legomena: words appearing only once
    hapax_count = sum(
        1 for count in word_counts.values() if count == 1
    )

    # Hapax ratio
    hapax_ratio = (
        hapax_count / total_words if total_words else 0
    )

    # Vocabulary richness
    vocabulary_richness = (
        unique_words / total_words if total_words else 0
    )

    results.append({
        "news_id": row.get("news_id", index),
        "label": row["label"],
        "total_words": total_words,
        "unique_words": unique_words,
        "TTR": ttr,
        "RTTR": rttr,
        "CTTR": cttr,
        "hapax_count": hapax_count,
        "hapax_ratio": hapax_ratio,
        "vocabulary_richness": vocabulary_richness
    })

# 4. Create results DataFrame
lexical_df = pd.DataFrame(results)

# 5. Display article-level results
print("Lexical Diversity Results:")
print(lexical_df.head())

# 6. Compare Real and Fake news
comparison = lexical_df.groupby("label")[
    [
        "total_words",
        "unique_words",
        "TTR",
        "RTTR",
        "CTTR",
        "hapax_count",
        "hapax_ratio",
        "vocabulary_richness"
    ]
].mean()

print("\nLexical Diversity Comparison: Real vs Fake News")
print(comparison.round(4))

# 7. Visualize lexical diversity metrics
metrics = ["TTR", "RTTR", "CTTR", "hapax_ratio"]

for metric in metrics:
    comparison[metric].plot(
        kind="bar",
        figsize=(7, 4)
    )

    plt.title(f"{metric} Comparison: Real vs Fake News")
    plt.xlabel("News Category")
    plt.ylabel("Average Score")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()

# 8. Save results
lexical_df.to_csv("T19_lexical_diversity.csv", index=False)
comparison.to_csv("T19_lexical_comparison.csv")

print("\nTask 19 completed successfully!")
print("Files saved:")
print("1. T19_lexical_diversity.csv")
print("2. T19_lexical_comparison.csv")
```
