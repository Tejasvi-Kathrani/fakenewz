```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer, KNNImputer 
from sklearn.experimental import enable_iterative_imputer 
from sklearn.impute import IterativeImputer 
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv("fakenews.csv")
print(df.isnull().sum())
```

```python
numeric_cols = df.select_dtypes(include=np.number).columns.tolist()
```

```python
if len(numeric_cols) == 0: 
    print("\nNo numeric columns available for imputation.") 
else:
     data = df[numeric_cols].copy()
```

```python
# Mean imputation 
mean_imputer = SimpleImputer(strategy="mean") 
mean_data = pd.DataFrame( mean_imputer.fit_transform(data), columns=numeric_cols )
```

```python
# Median imputation 
median_imputer = SimpleImputer(strategy="median") 
median_data = pd.DataFrame( median_imputer.fit_transform(data), columns=numeric_cols )
```

```python
# KNN imputation
knn_imputer = KNNImputer(n_neighbors=3) 
knn_data = pd.DataFrame( knn_imputer.fit_transform(data), columns=numeric_cols )
```

```python
# Iterative imputation
iterative_imputer = IterativeImputer(random_state=42) 
iterative_data = pd.DataFrame( iterative_imputer.fit_transform(data), columns=numeric_cols )
```

```python
print("\nMean Imputation:")
print(mean_data.head())
```

```python
print("\nMedian Imputation:") 
print(median_data.head())
```

```python
print("\nKNN Imputation:") 
print(knn_data.head())
```

```python
print("\nIterative Imputation:") 
print(iterative_data.head())
```

```python
# Check missing values after imputation
print("\nMissing Values After Imputation:")
print("Mean:", mean_data.isnull().sum().sum())  
print("Median:", median_data.isnull().sum().sum())
print("KNN:", knn_data.isnull().sum().sum())
print("Iterative:", iterative_data.isnull().sum().sum())
```
