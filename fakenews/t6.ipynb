```python
import pandas as pd
import re
df = pd.read_csv("fakenews.csv")
```

```python
# Function to standardize text
def standardize_text(text):
    if pd.isna(text):
        return ""

    # Convert text to lowercase
    text = text.lower()

    # Remove extra whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
# Standardize headline
df["headline"] = df["headline"].apply(standardize_text)

# Standardize body text
df["body_text"] = df["body_text"].apply(standardize_text)

# Display original and standardized text
print("Standardized Text Data:")
print(df[["headline", "body_text"]].head())

# Check for missing values
print("\nMissing Values After Standardization:")
print(df[["headline", "body_text"]].isnull().sum())
```

```python
# Save standardized dataset
df.to_csv("fakenews_standardized.csv", index=False)

print("\nText standardization completed successfully!")
print("Standardized dataset saved as fakenews_standardized.csv")
```
